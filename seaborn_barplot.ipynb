```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

1. Write a program to draw bar plot of sex against survived for a dataset given in the url
https://github.com/mwaskom/seaborn-data/blob/master/titanic.csv

```python
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv'

df = pd.read_csv(url)
grouped = df.groupby(['sex', 'survived']).size().unstack()
plt.figure(figsize=(8, 6))
grouped.plot(kind='bar', stacked=False, color=['darkblue', 'green'], ax=plt.gca())

plt.xlabel('Sex')
plt.ylabel('Count')
plt.title('Survival Count by Sex')
plt.legend(['Did Not Survive', 'Survived'], title='Survived')
plt.xticks(rotation=0)

plt.show()
```
