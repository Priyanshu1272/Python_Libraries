12.Write a Python program to create multiple types of charts

```python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([10, 20, 25, 30, 35])
y2 = np.array([5, 10, 15, 20, 25])
categories = ['A', 'B', 'C', 'D']
values = [25, 35, 20, 20]
hist_data = np.random.randn(1000)

plt.figure(figsize=(15, 10))

# 1. Line Chart
plt.subplot(2, 3, 1)  # 2 rows, 3 columns, position 1
plt.plot(x, y1, 'b-', label='Line 1')
plt.plot(x, y2, 'r--', label='Line 2')
plt.title('Line Chart')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)

# 2. Bar Chart
plt.subplot(2, 3, 2)
plt.bar(categories, values, color=['blue', 'green', 'red', 'purple'])
plt.title('Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

# 3. Scatter Chart
plt.subplot(2, 3, 3)
plt.scatter(x, y1, color='blue', label='Data 1')
plt.scatter(x, y2, color='red', label='Data 2')
plt.title('Scatter Chart')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()

# 4. Pie Chart
plt.subplot(2, 3, 4)
plt.pie(values, labels=categories, autopct='%1.1f%%', colors=['blue', 'green', 'red', 'purple'])
plt.title('Pie Chart')

# 5. Histogram
plt.subplot(2, 3, 5)
plt.hist(hist_data, bins=30, color='purple', edgecolor='black')
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
# Additional example: Box Plot
plt.figure(figsize=(8, 5))
data = [y1, y2, np.random.normal(20, 5, 5)]
plt.boxplot(data, labels=['Group 1', 'Group 2', 'Group 3'])
plt.title('Box Plot')
plt.ylabel('Values')
plt.show()
```

13.Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with linestyle -, width .5. and color blue. Date,Close 03-10-16,772.559998 04-10-16,776.429993 05-10-16,776.469971 06-10-16,776.859985 07-10-16,775.080017

```python
dates = ['03-10-16', '04-10-16', '05-10-16', '06-10-16', '07-10-16']
closing_values = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]

# Convert string dates to datetime objects
dates = [datetime.strptime(date, '%d-%m-%y') for date in dates]

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(dates, closing_values, 'b-o', label='Closing Value')

# Customize the grid
plt.grid(True, linestyle='-', linewidth=0.5, color='green')

# Customize the plot
plt.title('Alphabet Inc. Closing Values (Oct 3-7, 2016)')
plt.xlabel('Date')
plt.ylabel('Closing Value (USD)')
plt.legend()

# Format the x-axis dates
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d-%m-%y'))
plt.gca().xaxis.set_major_locator(mdates.DayLocator())
plt.xticks(rotation=45)

# Adjust layout to prevent label cutoff
#plt.tight_layout()

# Display the plot
plt.show()
```

14.Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with rendering with a larger grid (major grid) and a smaller grid (minor grid).Turn on the grid but turn off ticks.

```python
dates = ['03-10-16', '04-10-16', '05-10-16', '06-10-16', '07-10-16']
closing_values = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]

# Convert string dates to datetime objects
dates = [datetime.strptime(date, '%d-%m-%y') for date in dates]

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(dates, closing_values, 'b-o', label='Closing Value')
plt.grid(True, which='major', linestyle='-', linewidth=0.5, color='blue')
plt.grid(True, which='minor', linestyle='--', linewidth=0.3, color='lightblue')

# Turn off ticks
plt.tick_params(left=False, bottom=False, labelleft=True, labelbottom=True)

plt.title('Alphabet Inc. Closing Values (Oct 3-7, 2016)')
plt.xlabel('Date')
plt.ylabel('Closing Value (USD)')
plt.legend()

# Format the x-axis dates
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d-%m-%y'))
plt.gca().xaxis.set_major_locator(mdates.DayLocator())
plt.gca().xaxis.set_minor_locator(mdates.HourLocator(interval=12))  
plt.xticks(rotation=45)

# Set y-axis minor locator
plt.gca().yaxis.set_minor_locator(plt.MultipleLocator(0.5))  # Minor grid every 0.5 units

# Adjust layout
plt.tight_layout()
plt.show()
```

15.Write a Python program to create multiple plots.

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = x**2

# Approach 1: Multiple subplots in one figure (2x2 grid)
plt.figure(figsize=(12, 8))

# Subplot 1: Sine wave
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, position 1
plt.plot(x, y1, 'b-', label='Sine')
plt.title('Sine Wave')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)

# Subplot 2: Cosine wave
plt.subplot(2, 2, 2)
plt.plot(x, y2, 'r--', label='Cosine')
plt.title('Cosine Wave')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)

# Subplot 3: Tangent wave
plt.subplot(2, 2, 3)
plt.plot(x, y3, 'g-.', label='Tangent')
plt.title('Tangent Wave')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.ylim(-5, 5)  # Limit y-axis due to tangent's range

# Subplot 4: Quadratic function
plt.subplot(2, 2, 4)
plt.plot(x, y4, 'm:', label='Quadratic')
plt.title('Quadratic Function')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the first figure
plt.show()

# Approach 2: Multiple separate figures
# Figure 1: Sine and Cosine together
plt.figure(figsize=(8, 5))
plt.plot(x, y1, 'b-', label='Sine')
plt.plot(x, y2, 'r--', label='Cosine')
plt.title('Sine and Cosine Waves')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()

# Figure 2: Scatter plot of Quadratic
plt.figure(figsize=(8, 5))
plt.scatter(x, y4, color='purple', label='Quadratic', s=20)
plt.title('Quadratic Function (Scatter)')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```
