```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

1. Write a Python programming to display a bar chart of the popularity of programming
Languages.
Sample data:
Programming languages: Java, Python, PHP, JavaScript, C#, C++
Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7

```python
languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]
plt.figure(figsize=(10, 6)) 
plt.bar(languages, popularity, color='green')
plt.title('Popularity of Programming Languages', fontsize=14, pad=10)
plt.xlabel('Programming Languages', fontsize=12)
plt.ylabel('Popularity (%)', fontsize=12)

for i, v in enumerate(popularity):
    plt.text(i, v + 0.5, str(v), ha='center', fontsize=10)
plt.show()
```

10. Write a Python program to create bar plot of scores by group and gender. Use multiple X
values on the same chart for men and women.
Sample Data:
Means (men) = (22, 30, 35, 35, 26)
Means (women) = (25, 32, 30, 35, 29)

```python
groups = ['Group 1', 'Group 2', 'Group 3', 'Group 4', 'Group 5']
men_means = [22, 30, 35, 35, 26]
women_means = [25, 32, 30, 35, 29]
x = np.arange(len(groups))  # Creates array [0, 1, 2, 3, 4]
bar_width = 0.35  # Width of each bar
plt.figure(figsize=(10, 6))
plt.bar(x - bar_width/2, men_means, bar_width, label='Men', color='blue')
plt.bar(x + bar_width/2, women_means, bar_width, label='Women', color='pink')
plt.xlabel('Groups')
plt.ylabel('Scores')
plt.title('Scores by Group and Gender')
plt.xticks(x, groups)  
plt.legend()
```

11. Write a Python program to create bar plot from a DataFrame.
Sample Data Frame:
a b c d e
2 4,8,5,7,6
4 2,3,4,2,6
6 4,7,4,7,8
8 2,6,4,8,6
10 2,4,3,3,2

```python
data = {
    'a': [2, 4, 6, 8, 10],
    'b': [4, 2, 4, 2, 2],
    'c': [8, 3, 7, 6, 4],
    'd': [5, 4, 4, 4, 3],
    'e': [7, 2, 7, 8, 6]
}
df = pd.DataFrame(data)
plt.figure(figsize=(10, 6))
df.plot(kind='bar', ax=plt.gca())

plt.xlabel('Index')
plt.ylabel('Values')
plt.title('Bar Plot from DataFrame')
plt.legend(title='Columns')
plt.show()
```

12. Write a Python program to create bar plots with error bars on the same figure. Sample Date
Mean velocity: 0.2474, 0.1235, 0.1737, 0.1824
Standard deviation of velocity: 0.3314, 0.2278, 0.2836, 0.2645

```python
categories = ['A', 'B', 'C', 'D']
mean_velocity = [0.2474, 0.1235, 0.1737, 0.1824]
std_velocity = [0.3314, 0.2278, 0.2836, 0.2645]
# Set positions for bars
x = np.arange(len(categories))

plt.figure(figsize=(10, 6))
plt.bar(x, mean_velocity, yerr=std_velocity, capsize=5, color='skyblue')
plt.xlabel('Categories')
plt.ylabel('Mean Velocity')
plt.title('Bar Chart of Mean Velocity with Error Bars')
plt.xticks(x, categories)
plt.grid(True)
plt.show()
```

13. Write a Python program to create bar plots with errorbars on the same figure. Attach a text
label above each bar displaying men means (integer value).
Sample Data
Mean velocity: 0.2474, 0.1235, 0.1737, 0.1824
Standard deviation of velocity: 0.3314, 0.2278, 0.2836, 0.2645

```python
categories = ['A', 'B', 'C', 'D']
mean_velocity = [0.2474, 0.1235, 0.1737, 0.1824]
std_velocity = [0.3314, 0.2278, 0.2836, 0.2645]

x = np.arange(len(categories))
plt.figure(figsize=(10, 6))
bars = plt.bar(x,mean_velocity, yerr=std_velocity, capsize=5, color='skyblue')

for bar, mean in zip(bars, mean_velocity):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.01, f'{int(mean*100)}', 
             ha='center', va='bottom')

plt.xlabel('Categories')
plt.ylabel('Mean Velocity')
plt.title('Mean Velocity with Error Bars and Integer Labels')
plt.xticks(x, categories)

plt.show()
```
